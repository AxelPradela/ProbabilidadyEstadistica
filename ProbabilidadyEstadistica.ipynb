```
data=pd.read_csv("indices-bursatiles-argentina-indices-cotizaciones.csv")
```

```
data.head()
```

```
data["retorno"] = data["merval"].pct_change() # Crea una nueva columna llamada "retorno" que representa la variación porcentual diaria del índice MERVAL.
```

```
mu=data["retorno"].mean() # Calcula la media de los retornos.
mu
```

```
sigma = data["retorno"].std() # Calcula el desvío estándar de los retornos.
sigma
```

```
import numpy as np

x = data["retorno"].dropna()

sb.histplot(x, bins=40, stat="density")

plt.axvline(0.01, color="red")

sb.kdeplot(x, color="blue")

plt.title("Probabilidad de retorno > 1%")
plt.show()
```
