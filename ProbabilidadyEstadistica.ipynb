```
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.gridspec as gridspec
import seaborn as sb
import scipy as sc
```

```
data = {
    "team": ["Boca", "River", "Racing", "Independiente", "San Lorenzo"],
    "points": [45, 50, 38, 35, 40],
    "goals_for": [30, 42, 28, 25, 33]
}

df = pd.DataFrame(data)
df_sorted=df.sort_values(by="points",ascending=False)
df_sorted
```

```

plt.figure(figsize=(8,5))
plt.bar(df_sorted["team"],df_sorted["points"],color="skyblue")

plt.title("puntos por equipo")
plt.xlabel("Equipo")
plt.ylabel("Puntos")

plt.xticks(rotation=45)

plt.show()
```
